## Question 1.8 (4 pts)
Segment the dep_time 24 hours into 
* night from 12 midnight to 5 am
* morning from 5:01 am to 12 noon
* afternoon from 12.01 noon to 5 pm
* evening from 5.01 pm to 12 am

And then run a query to find the total flights in each segment and save that value in the given variable.


````python
def time_segment(dep_time):
    try:
        if 0 <= dep_time <= 500:
            return "night"
        elif 500 < dep_time <= 1200:
            return "morning"
        elif 1200 < dep_time <= 1700:
            return "afternoon"
        elif 1700 < dep_time <= 2400: # althought it is consistent with the test, but better to move 2400 to the night group
            return "evening"
        else:
            return "unknown"
    except:
        return "unknown"

df["time_segment"] = df["dep_time"].apply(time_segment)
parts_of_day_count = df["time_segment"].value_counts()
````

````python
assert parts_of_day_count['morning'] == 129939
assert parts_of_day_count['evening'] == 98362
````

## Question 2.1 (3 pts)

Line charts are ideal to show a trend line. Draw a line chart to show monthly frequency count of flights as a trend line for the year 2013. Be sure to choose the correct data to display on the x-axis and y-axis.

````python
import matplotlib.pyplot as plt
monthly_counts = flights[flights["year"] == 2013].groupby("month").size()
fig, ax = plt.subplots()
fig1 = ax
line = ax.plot(monthly_counts.index, monthly_counts.values)[0]

ax.set_title("Monthly breakdown")
ax.set_xlabel("Months (1-12)")
ax.set_ylabel("Monthly Frequency")
plt.show()
````

## Question 2.3 (3 pts)

Segment the flights based on time-slots with the following criteria
* flights that depart from 00-5:59am as 'red-eye' flights
* flights that depart from 6am - 11:59am as 'morning'
* flights that depart from 12noon - 3:59pm as 'afternoon'
* flights that depart from 4pm - 11:59pm as 'evening' 

And using seaborn create a plot as shown below. PS: Feel free to choose your color palette.

segmentation.png

````python
def time_segment(dep_time):
    try:
        if (0 <= dep_time <= 559) | (dep_time == 2400):
            return "red-eye"
        elif 600 <= dep_time <= 1159:
            return "morning"
        elif 1200 <= dep_time <= 1559:
            return "afternoon"
        elif 1600 <= dep_time <= 2359:
            return "evening"
        else:
            return "unknown"
    except:
        return "unknown"

flights["plot_time"] = flights["dep_time"].apply(time_segment)
grouped = flights.groupby(["plot_time", "origin"]).size().reset_index(name="count")
````

````python
import seaborn as sns

plot_order = ["red-eye", "morning", "afternoon", "evening"]
custom_titles = {
    "red-eye": "red-eye(00-5:59am)",
    "morning": "morning(6am-11:59am)",
    "afternoon": "afternoon(12noon-3:59pm)",
    "evening": "evening(4pm-11:59pm)"
}

fig_3 = sns.FacetGrid(grouped, col="plot_time", col_order=plot_order, sharey=True)
fig_3.map_dataframe(sns.barplot, x="origin", y="count", hue="origin",palette = "deep" )
for ax, title_key in zip(fig_3.axes.flat, plot_order):
    ax.set_title(custom_titles[title_key])
    ax.set_ylabel("Flight Count")

# Optional: adjust layout
plt.tight_layout()
plt.show()
````

````python
assert fig_3.figure.get_axes()[0].get_title() == 'red-eye(00-5:59am)'
assert fig_3.figure.get_axes()[0].get_ylabel() == 'Flight Count'
assert fig_3.figure.get_axes()[1].get_title() == 'morning(6am-11:59am)'
````
